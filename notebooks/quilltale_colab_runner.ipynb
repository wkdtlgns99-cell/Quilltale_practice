### 2단계: API 키 입력 및 3대 직업 병렬 자동 관전 시작
3개 계정의 Gemini API 키를 넣고 재생 버튼(▶)을 누르면 모험이 시작됩니다.

```python
#@title 🚀 2단계: 3계정 직업별 모험 및 합체 데이터 수집 실행

전사_API_KEY = "" #@param {type:"string"}
마법사_API_KEY = "" #@param {type:"string"}
도적_API_KEY = "" #@param {type:"string"}
각_직업별_턴수 = 50 #@param {type:"integer"}

import subprocess
import sys

cmd = [
    sys.executable, "scripts/batch_collector.py",
    "--key1", 전사_API_KEY,
    "--key2", 마법사_API_KEY,
    "--key3", 도적_API_KEY,
    "--turns", str(각_직업별_턴수)
]

print(f"🎮 3대 직업 (전사, 마법사, 도적) 각 {각_직업별_턴수}턴 모험을 시작합니다...\n")
subprocess.run(cmd, check=True)
print("\n🎉 모든 모험이 완료되었고 합체 학습 파일(data/dataset_master_combined.jsonl)이 생성되었습니다!")
```
