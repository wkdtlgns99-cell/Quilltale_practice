### 3단계 (핵심): 🧼 AI 데이터 세탁기 가동 (제미나이 명작 소설 윤문)
수집된 모험 사건 기록을 제미나이가 최고급 다크 판타지 소설 문체로 싹 다듬어줍니다.

```python
#@title 🧼 3단계: AI 세탁기 실행 (소설급 문체로 고품질 윤문)
세탁용_API_KEY = "" #@param {type:"string"}
세탁할_최대_턴수 = 50 #@param {type:"integer"}

import subprocess
import sys

key_to_use = 세탁용_API_KEY or 전사_API_KEY
cmd = [
    sys.executable, "scripts/data_refiner.py",
    "--input", "data/collector_logs/warrior_logs.json",
    "--output", "data/dataset_washed_master.jsonl",
    "--key", key_to_use,
    "--max", str(세탁할_최대_턴수)
]

print("🧼 AI 세탁기 가동 중... 문장을 소설가 필체로 변환합니다.\n")
subprocess.run(cmd, check=True)
print("\n🎉 최고급 명작 데이터셋(data/dataset_washed_master.jsonl) 생성 완료!")
```
